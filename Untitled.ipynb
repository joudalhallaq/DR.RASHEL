```python
print("Shape of dataset:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset info:")
df.info()

print("\nStatistical summary:")
display(df.describe(include='all'))
```

```python
missing_before = df.isnull().sum()
missing_before
duplicates_before = df.duplicated().sum()
print("Duplicate rows before cleaning:", duplicates_before)
print(df['TotalCharges'].head(10))
print(df['TotalCharges'].dtype)
```

### Observation
The `TotalCharges` column should be numeric, but it is stored as an object data type. This usually happens because some rows contain blank spaces instead of valid numeric values.

```python
df_clean = df.copy()
df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce')
df_clean = df_clean.dropna()
df_clean = df_clean.drop_duplicates()
df_clean.columns = df_clean.columns.str.strip()
```

```python
df_clean['ChargesPerMonth'] = df_clean['TotalCharges'] / df_clean['tenure']
df_clean['ChargesPerMonth'] = df_clean['ChargesPerMonth'].replace([np.inf, -np.inf], np.nan)
df_clean['ChargesPerMonth'] = df_clean['ChargesPerMonth'].fillna(df_clean['MonthlyCharges'])
```

```python
df_clean['Churn'] = df_clean['Churn'].str.strip()
```

```python
summary_table = pd.DataFrame({
    'Metric': ['Rows', 'Columns', 'Missing Values', 'Duplicate Rows'],
    'Before Cleaning': [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    'After Cleaning': [
        df_clean.shape[0],
        df_clean.shape[1],
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum()
    ]
})

summary_table
```

```python
print("Data types after cleaning:")
print(df_clean.dtypes)
```

### Cleaning Summary
The dataset was cleaned by converting `TotalCharges` to numeric format, removing rows with missing values, removing duplicate rows, and creating a new feature called `ChargesPerMonth`. These steps improved data consistency and made the dataset ready for visualization.

```python
tenure_counts = df_clean.groupby('tenure')['customerID'].count()

plt.figure(figsize=(10, 5))
plt.plot(tenure_counts.index, tenure_counts.values, marker='o', linewidth=2, label='Customer Count')
plt.title('Number of Customers by Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')
plt.grid(True)
plt.legend()

max_idx = tenure_counts.idxmax()
max_val = tenure_counts.max()
plt.annotate(f'Peak at {max_idx} months',
             xy=(max_idx, max_val),
             xytext=(max_idx+5, max_val+50),
             arrowprops=dict(arrowstyle='->'))

plt.show()
```

The line chart shows how customers are distributed across tenure values. A peak at specific tenure periods may indicate common subscription durations or retention patterns.

```python
contract_churn = pd.crosstab(df_clean['Contract'], df_clean['Churn'])

contract_churn.plot(kind='bar', figsize=(10, 6))
plt.title('Customer Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.grid(axis='y')

plt.annotate('Month-to-month contracts tend to have higher churn',
             xy=(0, contract_churn.loc['Month-to-month', 'Yes']),
             xytext=(0.4, contract_churn.values.max()*0.9),
             arrowprops=dict(arrowstyle='->'))

plt.show()
```

Customers on month-to-month contracts appear more likely to churn than customers on one-year or two-year contracts.

```python
plt.figure(figsize=(10, 6))
churn_colors = df_clean['Churn'].map({'Yes': 'red', 'No': 'blue'})

plt.scatter(df_clean['tenure'], df_clean['MonthlyCharges'], c=churn_colors, alpha=0.5)
plt.title('Tenure vs Monthly Charges')
plt.xlabel('Tenure (Months)')
plt.ylabel('Monthly Charges ($)')
plt.grid(True)

plt.annotate('Some short-tenure customers\nhave high monthly charges',
             xy=(5, 90), xytext=(15, 110),
             arrowprops=dict(arrowstyle='->'))

from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='Churn = Yes', markerfacecolor='red', markersize=8),
    Line2D([0], [0], marker='o', color='w', label='Churn = No', markerfacecolor='blue', markersize=8)
]
plt.legend(handles=legend_elements)

plt.show()
```

The scatter plot suggests that customers with short tenure and relatively high monthly charges may be more likely to churn.

```python
plt.figure(figsize=(10, 6))
plt.hist(df_clean['MonthlyCharges'], bins=30, edgecolor='black')
plt.title('Distribution of Monthly Charges')
plt.xlabel('Monthly Charges ($)')
plt.ylabel('Frequency')
plt.grid(axis='y')

mean_charge = df_clean['MonthlyCharges'].mean()
plt.axvline(mean_charge, linestyle='--', label=f'Mean = {mean_charge:.2f}')
plt.legend()

plt.annotate('Average monthly charge',
             xy=(mean_charge, 350),
             xytext=(mean_charge+10, 450),
             arrowprops=dict(arrowstyle='->'))

plt.show()
```

The histogram shows the spread of monthly charges and helps identify whether most customers pay low, medium, or high monthly amounts.

```python
churn_counts = df_clean['Churn'].value_counts()

plt.figure(figsize=(8, 8))
wedges, texts, autotexts = plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4)
)

plt.title('Customer Churn Distribution (Donut Chart)')
plt.annotate('Most customers did not churn',
             xy=(0.2, 0.6), xytext=(1.2, 1.0),
             arrowprops=dict(arrowstyle='->'))

plt.show()
```

The donut chart indicates that the majority of customers remained with the company, while a smaller portion churned.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# subplot 1
axes[0].hist(df_clean['tenure'], bins=20, edgecolor='black')
axes[0].set_title('Distribution of Tenure')
axes[0].set_xlabel('Tenure (Months)')
axes[0].set_ylabel('Frequency')
axes[0].grid(axis='y')

# subplot 2
internet_counts = df_clean['InternetService'].value_counts()
axes[1].bar(internet_counts.index, internet_counts.values)
axes[1].set_title('Customers by Internet Service Type')
axes[1].set_xlabel('Internet Service')
axes[1].set_ylabel('Number of Customers')
axes[1].grid(axis='y')

axes[1].annotate('Fiber optic and DSL dominate',
                 xy=(0, internet_counts.iloc[0]),
                 xytext=(0.3, internet_counts.max()+200),
                 arrowprops=dict(arrowstyle='->'))

plt.tight_layout()
plt.show()
```

The subplot layout combines two different chart types to compare customer tenure distribution with internet service subscription patterns.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_clean['TotalCharges'], bins=30, kde=True)
plt.title('Distribution of Total Charges')
plt.xlabel('Total Charges ($)')
plt.ylabel('Frequency')
plt.annotate('Most customers are concentrated\nin lower to middle total charges',
             xy=(1000, 500), xytext=(2500, 700),
             arrowprops=dict(arrowstyle='->'))
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df_clean, x='PaymentMethod', hue='Churn')
plt.title('Churn by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Count')
plt.xticks(rotation=20)
plt.annotate('Electronic check customers\nseem to churn more',
             xy=(0, 500), xytext=(1, 900),
             arrowprops=dict(arrowstyle='->'))
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_clean, x='tenure', y='TotalCharges', hue='Churn', alpha=0.6)
plt.title('Tenure vs Total Charges by Churn')
plt.xlabel('Tenure (Months)')
plt.ylabel('Total Charges ($)')
plt.annotate('Longer tenure naturally leads\nto higher total charges',
             xy=(50, 4000), xytext=(20, 5500),
             arrowprops=dict(arrowstyle='->'))
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_clean, x='Churn', y='MonthlyCharges')
plt.title('Monthly Charges by Churn Status')
plt.xlabel('Churn')
plt.ylabel('Monthly Charges ($)')
plt.annotate('Churned customers often have higher charges',
             xy=(1, 80), xytext=(0.4, 110),
             arrowprops=dict(arrowstyle='->'))
plt.show()
```

```python
df_corr = df_clean.copy()
df_corr['ChurnBinary'] = df_corr['Churn'].map({'Yes': 1, 'No': 0})

corr_cols = ['tenure', 'MonthlyCharges', 'TotalCharges', 'ChargesPerMonth', 'ChurnBinary']
corr_matrix = df_corr[corr_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

The heatmap helps identify numerical relationships, especially the strong connection between tenure and total charges.

```python
sns.pairplot(df_clean[['tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']], hue='Churn')
plt.show()
```

The pairplot provides a multi-variable view and highlights how churn differs across combinations of tenure, monthly charges, and total charges.

##Summary


This discussion has examined the Telco Customer Churn data to learn about customer behavior and the factors that are associated with churn. The data contains demographic details of the customers, subscription, billing and churn status. Once the data was loaded into Pandas, the first major quality problem that was found was that the column containing the TotalCharges was of type object rather than numeric. This problem was resolved by performing type conversion and deletion of rows that had no data. Duplicates were also verified and eliminated and another attribute was added known as ChargesPerMonth to facilitate the analysis.

The graphs showed some significant trends. The churn by contract type bar chart revealed that the churn of the customers in the month to month contracts was significantly higher as compared to the churn of the customers in the one year and two year contracts. This gives an indication that longer contracts could have a greater retention and less churn. The scatter plot of tenure and monthly charges also indicated that most of the customers that had a short tenure and those with significantly higher monthly charges were more likely to churn meaning that early-stage customers might be more price sensitive.

The histogram of monthly charges indicated that the customer charges are distributed over a large span with the focus being at medium-to-high billing. The donut chart of the churn distribution indicated that the number of churned customers is still high to warrant the business but most customers are still retained with the company. The churned customers in the Seaborn boxplot seem to be charged more on a monthly basis on an average, which substantiates the fact that pricing could be a significant factor of customer dissatisfaction or switching services.

The heatmap validated the strong positive relation between tenure and the total charges, which is understandable since longer-term customers will naturally have higher bills as they will be able to stay longer. The countplot of payment methods also indicated that the number of customers who use electronic checks to pay is more likely to churn than those who use other methods of payment.

On the basis of this analysis, the company ought to target customers with a monthly contract, high monthly fees, and low tenure. Potential suggestions may involve the provision of discounts to long-term contracts, enhancement of the onboarding of new customers, and restructuring of prices of the more likely to churn customers.
